```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['transaction_type'] = encoder.fit_transform(df['transaction_type'])
df['location_risk'] = encoder.fit_transform(df['location_risk'])
df['device_type'] = encoder.fit_transform(df['device_type'])
```

```python
df.head()
```

```python
X = df.drop('is_fraud', axis=1)
y = df['is_fraud']
```

```python
X.head()
y.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
    )
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train, y_train)
lr_pred = lr.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print("Logistic Regression Accuracy")
lr_accuracy = accuracy_score(y_test, lr_pred)
print(lr_accuracy)
```

```python
from sklearn.metrics import precision_score
print("Precision")
print(precision_score(y_test, lr_pred))
```

```python
from sklearn.metrics import recall_score
print("Recall")
print(recall_score(y_test, lr_pred))
```

```python
from sklearn.metrics import f1_score
print("F1 Score")
print(f1_score(y_test, lr_pred))
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, lr_pred)
print(cm)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, lr_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
dt_pred = dt.predict(X_test)
```

```python
dt_accuracy = accuracy_score(y_test, dt_pred)
print(dt_accuracy)
```

```python
print(precision_score(y_test, dt_pred))
```

```python
print(recall_score(y_test, dt_pred))
```

```python
print(f1_score(y_test, dt_pred))
```

```python
print(confusion_matrix(y_test, dt_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
rf_accuracy = accuracy_score(y_test, rf_pred)
print(f"Random Forest Accuracy:",rf_accuracy)
```

```python
from sklearn.metrics import precision_score
rf_precision = precision_score(y_test, rf_pred)
print(f"Random Forest Precision:",rf_precision)
```

```python
from sklearn.metrics import recall_score
rf_recall = recall_score(y_test, rf_pred)
print(f"Random Forest Recall:",rf_recall)
```

```python
from sklearn.metrics import f1_score
rf_f1 = f1_score(y_test, rf_pred)
print(f"Random Forest F1 Score:",rf_f1)
```

```python
from sklearn.metrics import confusion_matrix
rf_cm = confusion_matrix(y_test, rf_pred)
print("Random Forest Confusion Matrix:")
print(rf_cm)
```

```python
from sklearn.metrics import classification_report
print("Random Forest Classification Report:")
print(classification_report(y_test, rf_pred))
```
